```python
import pandas as pd

csv_path = "F:/cx da/data/oct.csv"
df = pd.read_csv(csv_path)

# Drop unneeded columns
cols_to_drop = ["Period", "End Date", "Time Increment", "Currency Code"]
df = df.drop(columns=cols_to_drop, errors="ignore")

# Parse to pandas datetime
df["Start Date"] = pd.to_datetime(df["Start Date"])

# Normalize names for the model
df = df.rename(
    columns={
        "Start Date": "date",
        "Sales": "sales",
        "Orders": "orders",
        "Ticket Size": "ticket_size",
    }
)

print(df.columns.tolist())
print(df.dtypes)
df.head(100)
```
